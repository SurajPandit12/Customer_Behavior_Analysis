```python
df['Review Rating'] = df['Review Rating'].fillna(df['Review Rating'].median())
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
```

```python
df.info()
```

```python
df.columns
```

```python
df['age_group'] = pd.cut(
    df['age'],
    bins=[0, 25, 35, 50, float('inf')],
    labels=['young', 'adult', 'middle_age', 'old']
)
```

```python
df[['age', 'age_group']].head(10)
```

```python
frequency_days = {
    'Weekly': 7,
    'Fortnightly': 14,
    'Monthly': 30,
    'Quarterly': 90,
    'Annually': 365
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_days)
```

```python
df[['frequency_of_purchases', 'purchase_frequency_days']].head(10)
```

```python
from sqlalchemy import create_engine

# Connect
engine = create_engine(
    "postgresql+psycopg2://postgres:1234@localhost:5432/customer_behaviour"
)

# Upload DataFrame
df.to_sql(
    'customer_data',
    engine,
    if_exists='replace',
    index=False
)
print(f"Data successfully loaded")
```
